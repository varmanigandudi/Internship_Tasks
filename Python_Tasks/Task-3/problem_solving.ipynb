Q1. Find the Frequency of Each Word in a Sentence
Write a Python program to count how many times each word appears in a given sentence and store the result in a dictionary.
Example Input:
python is easy and python is powerful
Expected Output:
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
#Psedocode
def frequency(sen):
    sen=sen.split()
    new_dict={}
    for word in sen:
     count=sen.count(word)
     new_dict[word]=count
    return new_dict
1.create function frequency with positional arugument sen
2.split sen into word using function sen.split()
3.create a new dictionary
4.using for loop iterate through the sen
5.count each word in the sentence
6.add a word and count to new dict
7.retun a new dict

```python
def frequency(sen):
    sen=sen.split()
    new_dict={}
    for word in sen:
     count=sen.count(word)
     new_dict[word]=count
    return new_dict
x=frequency('kohli is best batsmen')
print(x)
```
