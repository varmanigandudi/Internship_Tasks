```python
'''Q1. Find the Shortest Word in a Sentence
Write a Python program to find the shortest word in a given sentence.
Example input:
Python makes coding easy
Expected output:
easy'''
sen=input().split()
shortest_word=sen[0]
for word in sen:
  if len(word)<len(shortest_word):
    shortest_word=word
print("shortest_word:",shortest_word)
```

```python
'''Q2. Find the Largest Number in a List
Write a Python program to find the largest number in a list without using the built-in max() function.
Example input:
[12, 45, 7, 89, 23]
Expected output:
89'''
nums=[12, 45, 7, 89, 23]
largest_number=0
for i in nums:
  if i>largest_number:
    largest_number=i
print("largest_number:",largest_number)
```

```python
'''Q3. Find the Smallest Number in a List
Write a Python program to find the smallest number in a list without using the built-in min() function.
Example input:
[12, 45, 7, 89, 23]
Expected output:
7'''
nums=[12,45,7,89,23]
smallest_number=nums[0]
for i in nums:
  if i<smallest_number:
    smallest_number=i
print("smallest number:",smallest_number)
```

```python
'''Q4. Find the First Longest Word When Ties Occur
Write a Python program to find the first longest word in a sentence. If multiple words have the same maximum length, return the one that appears first.
Example input:
Python makes coding useful
Expected output:
Python'''
sen=input().split()
longest_word=""
for word in sen:
  if len(word)>len(longest_word):
    longest_word=word
    break
print("First longest word:",longest_word)
```

```python
'''Q5. Find the Word with the Maximum Number of Characters
Write a Python program to identify the word containing the maximum number of characters in a sentence.
Example input:
Data engineering requires problem solving
Expected output:
engineering'''
sen=input().split()
max_char=""
for word in sen:
  if len(word)>len(max_char):
    max_char=word
print("maximum number of characters are:",max_char)
```

```python
'''Q6. Count Vowels in a String
Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.
Example input:
Education
Expected output:
5'''
word=input().lower()
vowel_count=0
for ch in word:
  if ch in 'aeiou':
    vowel_count+=1

print("vowel count:",vowel_count)
```

```python
'''Q7. Count Consonants in a String
Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.
Example input:
Python 123!
Expected output:
5'''
word=input().lower().replace(" ","")
consonant_count=0
has_special_char=False
has_digit=False
for ch in word:
  if ch not in 'aeiou' and ch.isalpha():
    consonant_count+=1

print("consonant count:",consonant_count)

```

```python
'''Q8. Find the Most Frequent Character
Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
Example input:
banana
Expected output:
a'''
word=input()
for ch in word:
  count=word.count(ch)
  if count>1:
    break
print(ch)
```

```python
'''Q9. Find the Least Frequent Character
Write a Python program to find the character that appears least frequently in a string. If multiple characters have the same lowest frequency, return the first one encountered. Ignore spaces.
Example input:
success
Expected output:
u'''
word=input()
for ch in word:
  count=word.count(ch)
  if count==1:
    break
print(ch)
```

```python
'''Q10. Find All Characters Appearing More Than Once
Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.
Example input:
programming
Expected output:
['r', 'g', 'm']'''
word=input()
seen=[]
dup=[]
for ch in word:
  if ch not in seen:
    seen.append(ch)
  else:
    dup.append(ch)
print(dup)
```

```python
'''Q11. Create a Dictionary of Words and Their First Characters
Write a Python program to create a dictionary in which each word is a key and its first character is the value.
Example input:
Python Data Engineering
Expected output:
{'Python': 'P', 'Data': 'D', 'Engineering': 'E'}'''
sen=input().split()
new_dict={}
for word in sen:
    new_dict[word]=word[0]
print(new_dict)
```

```python
'''Q12. Create a Dictionary of Words and Their Last Characters
Write a Python program to create a dictionary in which each word is a key and its last character is the value.
Example input:
Python Data Engineering
Expected output:
{'Python': 'n', 'Data': 'a', 'Engineering': 'g'}'''
sen=input().split()
new_dict={}
for word in sen:
    new_dict[word]=word[-1]
print(new_dict)
```

```python
'''Q13. Create a Dictionary of Words and Their Uppercase Versions
Write a Python program to create a dictionary in which each word is a key and its uppercase version is the value.
Example input:
data science python
Expected output:
{'data': 'DATA', 'science': 'SCIENCE', 'python': 'PYTHON'}'''
sen=input().split()
new_dict={}
for word in sen:
  new_dict[word]=word.upper()
print(new_dict)
```

```python
'''Q14. Create a Dictionary of Numbers and Their Squares
Write a Python program to create a dictionary containing each number in a list as a key and its square as the value.
Example input:
[2, 3, 4, 5]
Expected output:
{2: 4, 3: 9, 4: 16, 5: 25}'''
nums=[2,3,4,5]
square_dict={}
for i in nums:
  square_dict[i]=i**2
print(square_dict)
```

```python
'''Q15. Create a Dictionary of Numbers and Their Even/Odd Status
Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.
Example input:
[1, 2, 3, 4, 5]
Expected output:
{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}'''
nums=[1,2,3,4,5]
new_dict={}
for i in nums:
  if i%2==0:
    new_dict[i]='even'
  else:
    new_dict[i]='Odd'
print(new_dict)
```

```python
'''Q20. Find the Longest Word That Appears Only Once
Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.
Example input:
Python makes coding useful Python
Expected output:
coding'''
sen=input().lower().split()
seen=[]
dup=[]
longest_word=""
for word in sen:
  if word not in seen:
    seen.append(word)
  else:
    if word not in dup:
      dup.append(word)
for word in seen:
  if len(word)>len(longest_word) and word not in dup:
    longest_word=word
print(longest_word)
```
